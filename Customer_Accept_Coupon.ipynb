**Context**

Imagine driving through town and a coupon is delivered to your cell phone for a restaurant near where you are driving. Would you accept that coupon and take a short detour to the restaurant? Would you accept the coupon but use it on a subsequent trip? Would you ignore the coupon entirely? What if the coupon was for a bar instead of a restaurant? What about a coffee house? There are five different types of coupons -- less expensive restaurants (under \$20), coffee houses, carry out & take away, bar, and more expensive restaurants (\$20 - $50).

```python
# Chi-Square Test with Multiple Variables to chk whether one categorical variable is dependent on multiple other variables.  If p-value < 0.05, the predictor is likely associated with the target
# Bar~weather & temperature, CoffeeHouse ~ weather & temperature are likely linked
# Chk for categorical_columns = ['CoffeeHouse','Restaurant20To50','CarryAway','RestaurantLessThan20','Bar'] which have Null values
from scipy.stats import chi2_contingency

for col in ['passanger','weather','temperature','gender','age','maritalStatus','occupation','has_children','education','occupation','income']:  # predictor columns
    table = pd.crosstab(df[col], df['Bar'])
    chi2, p, dof, _ = chi2_contingency(table)
    print(f"{col} vs Target: p-value = {p:.4f}")
```

3. Decide what to do about your missing data -- drop, replace, other...

```python
# Imputation of Null Values with Mode for the columns with Null

columns_with_Nulls = ['CoffeeHouse','Restaurant20To50','CarryAway','RestaurantLessThan20','Bar']
for i in columns_with_Nulls:
  print(data[i].isnull().sum())
  data[i] = data[i].transform(lambda x: x.fillna(x.mode()[0] if not x.mode().empty else 'Unknown'))
  print(data[i].isnull().sum())

# Encoding categorical variables
from sklearn.preprocessing import LabelEncoder
for i in ['CoffeeHouse','Restaurant20To50','CarryAway','RestaurantLessThan20','Bar']:
  le = LabelEncoder()
  data[str(i)+'_encoded'] = le.fit_transform(data[i])

data.select_dtypes(include=['float64', 'int64']).apply(lambda x: x.unique())
```
